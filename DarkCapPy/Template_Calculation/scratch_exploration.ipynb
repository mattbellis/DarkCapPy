```python
import matplotlib.pylab as plt
import numpy as np
```

```python
x = df['M_DM']
y0 = df['frac_ecut10']
y1 = df['frac_hit_id_ecut10']

y2 = df['frac_ecut100']
y3 = df['frac_hit_id_ecut100']


plt.plot(x,y1/y0,'o', label='ratio hit ID/no ID (10 GeV)')
plt.plot(x,y3/y2,'o', label='ratio hit ID/no ID (100 GeV)')

plt.ylim(0.01, 0.04)
plt.xlabel(r'M$_{X}$ (GeV/c$^2$)',fontsize=14)
plt.ylabel('acceptance hit ID/acceptance trigger')
plt.xscale('log')

plt.legend()

#plt.savefig('ratio_of_efficiencies_core.png')
plt.savefig('ratio_of_efficiencies_floating.png')
```

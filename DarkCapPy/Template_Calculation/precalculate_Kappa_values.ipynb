```python
alpha = 1./137


#mydict = {"mx": [], "kappa0": [], "cCap": [], "cCapQuick": []}
mydict = {"mx": [], "kappa0": []}

for mx in mxs:

    mydict['mx'].append(mx)
    
    start = time.time()
    print(f"Starting kappa0... for mX = {mx}")    
    kappa0 = DP.kappa_0(mx, alpha)

    mydict['kappa0'].append(kappa0)

    print(f"Finished calculating Kappa_0: {kappa0:7.5e} in {time.time() - start:.2f} seconds")

print(mydict)

df = pd.DataFrame.from_dict(mydict)

######################################################################################
# Write to a file and try to encode some information in the file
# 
# Up to you to decide how to name the file
#
######################################################################################

#df.to_parquet('different_kappas_TEST_SUBTESTS_SO_I_DONT_OVERWRITE.parquet')
#df.to_parquet(f'different_kappas_mx_{mxlo}_{mxhi}_mxstep_{mxstep}.parquet')
#df.to_parquet(f'different_kappas_mx_{mxlo}TeV_{mxhi}TeV_mxstep_{mxstep}TeV.parquet')
#df.to_parquet(f'different_kappas_mx_{mxlo}GeV_{mxhi}GeV_mxstep_{mxstep}_log_GeV.parquet')

df.to_parquet(f'KAPPAS_FOR_TESTING.parquet')


df

```

```python
# From here
# https://stackoverflow.com/questions/35213402/make-a-linear-within-logspace-array-that-goes-like-1-2-3-10-20-30

def linlog(start, stop, num):
    
    a = np.arange(1, 10)
    b = np.logspace(start, stop, num=num)
    result = (a[np.newaxis, :] * b[:, np.newaxis]).flatten()
    return result

linlog(-11, -7, 5)
```

```python
# Dark photon lifetime

mas = 0.22

#ma = 0.25 # GeV
#epsilon = 1e-8
#epsilon = np.array([1e-11, 1e-10, 1e-9, 1e-8, 1e-7, 1e-6])
epsilon = linlog(-11,-7, 5)



br = 1.0

# To see what happens when there is no boost from the high mass of the parent particle
mx = mas

# Returns the decay length in centimeters
L = DP.decayLength(mx, mas, epsilon, br)

# Convert to meters
L /= 100

#print(f'{L =   :.2f} m')

# Then I think we can divide by the speed of light to get the lifetime
c = 3e8

tau = L/c

#print(f'{tau = :.2e} s')
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
plt.plot(epsilon, L)
plt.yscale('log')
plt.xscale('log')
#plt.ylim(0.08, 225)
lb = '{'
rb = '}'
plt.xlabel(f"$\epsilon$", fontsize=18)
plt.ylabel(f"c$\\tau$ [m]", fontsize=18)

plt.subplot(1,2,2)
plt.plot(epsilon, tau)
plt.yscale('log')
plt.xscale('log')
lb = '{'
rb = '}'
plt.xlabel(f"$\epsilon$", fontsize=18)
plt.ylabel(f"$\\tau$ [s]", fontsize=18)

plt.tight_layout()

dicttmp = {'epsilon':epsilon, 'tau':tau, 'L': L, 'ma':mas*np.ones_like(epsilon)}
df = pd.DataFrame.from_dict(dicttmp)

plt.savefig('lifetime_as_function_of_epsilon.png')

outfilename = 'lifetimes_by_eps_mA.parquet'
df.to_parquet(outfilename)

df
```
